```python
# Definisi fungsi non-linear f(x) yang akan dicari akarnya
# Contoh: f(x) = x^3 - x - 2
def f(x):
    return x**3 - x - 2

# Algoritma Metode Bisection (Bagi Dua)
def bisection(a, b, tol=1e-4, maks_iter=100):
    # Cek apakah akar ada di dalam interval [a, b]
    if f(a) * f(b) >= 0:
        print("Metode bisection gagal: f(a) dan f(b) harus berlawanan tanda.")
        return None

    print(f"{'Iterasi':<8} | {'a':<10} | {'b':<10} | {'c (Tengah)':<10} | {'f(c)':<10}")
    print("-" * 55)

    c = a
    for i in range(1, maks_iter + 1):
        c = (a + b) / 2.0

        # Cetak proses iterasi
        print(f"{i:<8} | {a:<10.4f} | {b:<10.4f} | {c:<10.4f} | {f(c):<10.4f}")

        # Cek jika nilai f(c) sudah sangat dekat dengan 0 atau interval sudah cukup kecil
        if f(c) == 0 or (b - a) / 2 < tol:
            break

        # Perbarui batas interval
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    return c

# Menentukan batas awal interval [1, 2]
a_batas = 1.0
b_batas = 2.0

# Menjalankan fungsi bisection
akar_persamaan = bisection(a_batas, b_batas)

if akar_persamaan is not None:
    print(f"\nAkar persamaan numerik ditemukan pada x = {akar_persamaan:.4f}")
```
