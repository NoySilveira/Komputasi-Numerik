```python
import numpy as np
print("NumPy siap!")

A = np.array([[2, 1, -1], [-3, -1, 2], [-2, 1, 2]], dtype=float)
b = np.array([8, -11, -3], dtype=float)

print("Matriks A (Koefisien):\n", A)
print("Vektor b (Konstanta):", b)

solusi = np.linalg.solve(A, b)

print(f"Hasil Penyelesaian: x = {solusi[0]:.4f}, y = {solusi[1]:.4f}, z = {solusi[2]:.4f}")

verifikasi = np.dot(A, solusi)
print("Verifikasi (A dikali solusi):", verifikasi)

print("Apakah hasil akurat dan valid?", np.allclose(verifikasi, b))
```
