```python
import numpy as np
print("NumPy siap!")

# Matriks koefisien A dan vektor b untuk SPL 3 variabel
A = np.array([[2, 1, -1], [-3, -1, 2], [-2, 1, 2]], dtype=float)
b = np.array([8, -11, -3], dtype=float)
print("Matriks A:\n", A)
print("Vektor b:", b)

solusi = np.linalg.solve(A, b)
print(f"Hasil: x = {solusi[0]:.4f}, y = {solusi[1]:.4f}, z = {solusi[2]:.4f}")

verifikasi = np.dot(A, solusi)
print("Verifikasi (A @ solusi):", verifikasi)
print("Apakah akurat?", np.allclose(verifikasi, b))
```
