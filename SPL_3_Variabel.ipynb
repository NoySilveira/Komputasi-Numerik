```python
# ==========================================
# PROGRAM PENYELESAIAN SPL 3 VARIABEL
# Mata Kuliah: Komputasi Numerik
# ==========================================

# 1. Mengimpor pustaka NumPy yang digunakan untuk operasi matriks dan aljabar linier
import numpy as np
print("NumPy siap!")

# 2. Mendefinisikan Matriks Koefisien (A) dan Vektor Hasil (b) untuk Persamaan Linear 3 Variabel
# Bentuk persamaan:
# 2x + y - z = 8
# -3x - y + 2z = -11
# -2x + y + 2z = -3
A = np.array([[2, 1, -1], [-3, -1, 2], [-2, 1, 2]], dtype=float)
b = np.array([8, -11, -3], dtype=float)

# Menampilkan matriks A dan vektor b ke layar
print("Matriks A (Koefisien):\n", A)
print("Vektor b (Konstanta):", b)

# 3. Menyelesaikan Sistem Persamaan Linear menggunakan fungsi bawaan numpy.linalg.solve()
solusi = np.linalg.solve(A, b)

# Menampilkan hasil perhitungan nilai x, y, dan z dengan format 4 angka di belakang koma
print(f"Hasil Penyelesaian: x = {solusi[0]:.4f}, y = {solusi[1]:.4f}, z = {solusi[2]:.4f}")

# 4. Verifikasi keakuratan hasil dengan mengalikan matriks A dengan vektor solusi (A @ solusi)
verifikasi = np.dot(A, solusi)
print("Verifikasi (A dikali solusi):", verifikasi)

# Mengecek apakah hasil perkalian mendekati atau sama persis dengan vektor b awal
print("Apakah hasil akurat dan valid?", np.allclose(verifikasi, b))
```
